# CampusBites - Data Inspection

## Project Objective

To inspect and understand the menu datasets
collected from six food outlets at UPES.

## Datasets

1. Annapurna
2. Appetite Rolls
3. Chai Garam
4. Food Court
5. Frisco
6. Tulips

```python
import pandas as pd
from pathlib import Path

# Define the data folder
DATA_PATH = Path("../data/raw")

# Function to load CSV files
def load_csv(filename):
    file_path = DATA_PATH / filename

    try:
        return pd.read_csv(file_path, encoding="utf-8-sig")
    except UnicodeDecodeError:
        return pd.read_csv(file_path, encoding="cp1252")


# Load all six datasets
annapurna = load_csv("annapurna_menu.csv")
appetite = load_csv("appetiterolls_menu.csv")
chai_garam = load_csv("chaigaram_menu.csv")
food_court = load_csv("foodcourt_menu.csv")
frisco = load_csv("frisco_menu.csv")
tulips = load_csv("tulips_menu.csv")

print("All six datasets loaded successfully!")
```

```python
# Creating a dictionary of all six datasets

datasets = {
    "Annapurna": annapurna,
    "Appetite Rolls": appetite,
    "Chai Garam": chai_garam,
    "Food Court": food_court,
    "Frisco": frisco,
    "Tulips": tulips
}

# Checking number of rows and columns

for shop, df in datasets.items():
    print(f"{shop}: {df.shape}")
```

```python
# Checking column names

for shop, df in datasets.items():
    print(f"\n{shop}")
    print(df.columns.tolist())
```

```python
# Displaying first five records from each shop

for shop, df in datasets.items():
    print(f"\n{'='*40}")
    print(shop)
    display(df.head())
```

```python
# STEP 5.1 - Missing Values Analysis

print("MISSING VALUES REPORT")

for shop, df in datasets.items():
    print(f"\n--- {shop} ---")
    print(df.isnull().sum())
```

```python
# Checking duplicate records in all datasets

print("DUPLICATE RECORDS REPORT")

for shop, df in datasets.items():
    duplicates = df.duplicated().sum()

    print(f"{shop}: {duplicates} duplicate records")
```

```python
# STEP 5.3 - Data Type Inspection

print("DATA TYPES REPORT")

for shop, df in datasets.items():
    print(f"\n--- {shop} ---")
    print(df.dtypes)
```

```python
# Checking whether price columns are numeric

for shop, df in datasets.items():
    print(
        f"{shop}:",
        pd.api.types.is_numeric_dtype(df["Price"])
    )
```

```python
# STEP 5.4 - Price Validation

print("PRICE VALIDATION REPORT")

for shop, df in datasets.items():

    print(f"\n--- {shop} ---")

    print("Minimum Price:", df["Price"].min())
    print("Maximum Price:", df["Price"].max())
    print("Average Price:", round(df["Price"].mean(), 2))
```

```python
# Checking zero and negative prices

for shop, df in datasets.items():

    zero_prices = (df["Price"] == 0).sum()
    negative_prices = (df["Price"] < 0).sum()

    print(f"\n{shop}")
    print("Zero prices:", zero_prices)
    print("Negative prices:", negative_prices)
```

```python
# Detecting price outliers using IQR

for shop, df in datasets.items():

    Q1 = df["Price"].quantile(0.25)
    Q3 = df["Price"].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df["Price"] < lower_limit) |
        (df["Price"] > upper_limit)
    ]

    print(f"\n--- {shop} ---")
    print("Potential price outliers:", len(outliers))
    display(outliers)
```

# STEP 6 – DATA CLEANING AND STANDARDIZATION

Cleaning and standardizing all six shop datasets
while preserving the original raw datasets.

```python
# STEP 6.1 - Create cleaned copies of datasets

import re
from pathlib import Path

processed_datasets = {}

# Create processed folder
PROCESSED_PATH = Path("../data/processed")
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)

for shop, df in datasets.items():

    # Create a copy
    clean_df = df.copy()

    # Clean Shop Name
    clean_df["Shop Name"] = (
        clean_df["Shop Name"]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    # Clean Item Names
    clean_df["Item to Buy"] = (
        clean_df["Item to Buy"]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    # Ensure Price is numeric
    clean_df["Price"] = pd.to_numeric(
        clean_df["Price"], errors="coerce"
    )

    # Store cleaned dataset
    processed_datasets[shop] = clean_df

    # Save cleaned CSV
    filename = shop.lower().replace(" ", "") + "_menu.csv"
    clean_df.to_csv(PROCESSED_PATH / filename, index=False)

    print(f"{shop}: Cleaning completed")
    print("Rows:", len(clean_df))
    print("Columns:", list(clean_df.columns))
    print("-" * 40)

print("All six cleaned datasets saved successfully!")
```

```python
# Verify cleaned datasets

for shop, df in processed_datasets.items():

    print(f"\n--- {shop} ---")

    print("Missing values:")
    print(df.isnull().sum())

    print("Data types:")
    print(df.dtypes)

    display(df.head(3))
```

# STEP 7 – FOOD CATEGORY CLASSIFICATION

Assigning categories to food items across all six shops.

```python
def assign_category(item):

    item = str(item).lower().strip()

    # 1. Beverages
    if any(word in item for word in [
        "coffee", "tea", "chai", "mocha", "cappuccino",
        "americano", "latte", "espresso", "darjeeling",
        "jasmine green", "chamomile", "mojito", "lemonade",
        "nimbu pani", "nimbupani", "blue heaven",
        "strawberry cooler", "sparkle", "juice",
        "shake", "lassi", "frappe", "soda", "slush",
        "falooda", "cold drink", "green tea", "kulhad tea",
        "kulhad chai", "masala chai", "hot chocolate"
    ]):
        return "Beverages"

    # 2. Desserts
    elif any(word in item for word in [
        "ice cream", "cake", "brownie", "pastry",
        "muffin", "chocolate bliss", "chocolate excess",
        "gulab jamun", "sundae", "truffle", "chocobar",
        "choco bar", "chocolate cone", "berry",
        "american nuts", "butterscotch", "choco chips",
        "double belgian", "alphonso mango", "nutty belgian",
        "chocolate tub", "zulu bar", "wild berry",
        "tutti fruity"
    ]):
        return "Desserts"

    # 3. South Indian
    elif any(word in item for word in [
        "dosa", "idli", "vada", "wada", "uttapam",
        "uttappam", "pesarattu", "sambar", "shambhar",
        "south indian"
    ]):
        return "South Indian"

    # 4. Main Course
    elif any(word in item for word in [
        "thali", "biryani", "rice", "curry", "naan",
        "roti", "paratha", "rajma", "chole bhatura",
        "dal", "poori", "kulcha", "omelette rice",
        "paneer bowl", "chicken bowl", "parantha"
    ]):
        return "Main Course"

    # 5. Fast Food
    elif any(word in item for word in [
        "pasta", "noodles", "pizza", "burger",
        "wrap", "roll", "sandwich", "maggi",
        "fries", "nuggets", "toast", "hot dog",
        "hotdog", "macaroni", "paneer twister",
        "cheese garlic", "peanut butter",
        "tandoori paneer", "chilly paneer"
    ]):
        return "Fast Food"

    # 6. Snacks
    elif any(word in item for word in [
        "samosa", "pakoda", "pakora", "poha",
        "patty", "patties", "chaat", "corn",
        "bread roll", "kachori", "omelette",
        "omelet", "bhurji", "half fry",
        "finger", "soup", "oats", "bread butter"
    ]):
        return "Snacks"

    # 7. Other
    else:
        return "Other"
```

```python
for shop, df in processed_datasets.items():

    df["Category"] = df["Item to Buy"].apply(assign_category)

    filename = shop.lower().replace(" ", "") + "_menu.csv"

    df.to_csv(PROCESSED_PATH / filename, index=False)

    print(f"{shop}: Updated successfully!")
```

```python
# Manually correcting remaining food categories

manual_categories = {
    "Annapurna": {
        "Plain Uthappam": "South Indian",
        "Onion Uthappam": "South Indian",
        "Tomato Uthappam": "South Indian",
        "Onion Tomato Uthappam": "South Indian",
        "Mix Uthappam": "South Indian"
    },

    "Chai Garam": {
        "Tomato Cheese": "Fast Food",
        "Hot n Sour": "Soups",
        "Thick Tomato": "Soups"
    },

    "Food Court": {
        "Sambhar": "South Indian",
        "Pindi Chole Bhature": "Main Course",
        "Pav Bhaji": "Main Course"
    },

    "Frisco": {
        "Death by Chocolate": "Desserts",
        "Pastries": "Desserts",
        "Flat White": "Beverages"
    },

    "Tulips": {
        "Triple Chocolate": "Desserts",
        "Choco Chip (Cup)": "Desserts",
        "Gauva Chilly (Tub)": "Desserts"
    }
}

# Apply manual classifications

for shop, categories in manual_categories.items():

    df = processed_datasets[shop]

    for item, category in categories.items():
        df.loc[df["Item to Buy"] == item, "Category"] = category

print("Manual category corrections completed!")
```
